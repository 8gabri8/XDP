```python
import pandas as pd

df = pd.read_excel("HD_phaseGenes.xlsx", sheet_name="Table S2 - Phase C,D Gene Stats", header=26)
print(df.Phase.value_counts())
df
```

```python
# Phase D = the 107 de-repressed genes (any Phase value), not Phase == "D"
phase_d_derep = df[df["De-repressed phase D"] == "Y"]
display(phase_d_derep[
    ["Gene", "Phase D p-value", "Phase D effect", "Phase", "De-repressed phase D", "Case-control fold-change"]
    ].sort_values("Case-control fold-change").tail(20))
print(phase_d_derep.Gene.to_list())

phase_d_derep.to_csv("phase_d_genes.csv", index=False)
```

```python
# phase c+
phase_c = df[(df["Phase"] == "C+") & (df["Highly expressed C+"] == "Y")]
display(phase_c[["Gene", "Phase C p-value", "Phase C effect", "Phase"]])
print(phase_c.Gene.to_list())

phase_c.to_csv("phase_c_plus_genes.csv", index=False)
```

```python
# phase c-
phase_c_minus = df[(df["Phase"] == "C-") & (df["Highly expressed C-"] == "Y")]
display(phase_c_minus[["Gene", "Phase C p-value", "Phase C effect", "Phase"]])
print(phase_c_minus.Gene.to_list())

phase_c_minus.to_csv("phase_c_minus_genes.csv", index=False)
```
